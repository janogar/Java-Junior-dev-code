## 1.2 Álgebra Lineal Aplicada

Se crean arrays de cantidades y precios a partir del dataset y se calcula el ingreso bruto como producto punto.

```python

csv_path = 'ventas_techventas.csv'

# Si el problema de las comillas era a nivel de línea completa, 
# se puede limpiar antes de pasarlo a pandas de forma correcta:
fixed_lines = []
with open(csv_path, 'r', encoding='latin-1') as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        # Corregir si la línea entera está encapsulada en comillas dobles
        if line.startswith('"') and line.endswith('"'):
            line = line[1:-1]
        fixed_lines.append(line)

df = pd.read_csv(StringIO('\n'.join(fixed_lines)), parse_dates=['fecha'])

# Asegurar tipos correctos
df['cantidad'] = df['cantidad'].astype(int)
df['precio_unitario'] = df['precio_unitario'].astype(float)

cantidades = df['cantidad'].to_numpy()
precios = df['precio_unitario'].to_numpy()

ingresos_dot = np.dot(cantidades, precios)
ingresos_elem = cantidades * precios

ingresos_dot, ingresos_elem[:10]
```

### Comparación de multiplicaciones

- `np.dot(cantidades, precios)` da el ingreso total bruto del periodo.
- `cantidades * precios` produce ingresos por fila que luego se pueden sumar.

```python
ingresos_elem_sum = ingresos_elem.sum()
ingresos_dot, ingresos_elem_sum
```

## 1.3 Estadísticas con NumPy

Calculamos media, mediana, desviación estándar y percentiles del precio unitario.

```python
media_precio = np.mean(precios)
mediana_precio = np.median(precios)
std_precio = np.std(precios, ddof=0)
percentiles_precio = np.percentile(precios, [25, 75])
media_precio, mediana_precio, std_precio, percentiles_precio
```
